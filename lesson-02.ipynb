## 3｜為數值取名字：變數
當價格和數量會改變，用有意義的名字記錄它們，比反覆修改算式更容易閱讀。

```python
price = 30
```

這裡是把右邊的數值 `30` 指定給左邊的變數 `price`。**`=` 是賦值，不是在詢問兩邊是否相等。**

`price` 表示單價，`quantity` 表示數量，`total` 表示總金額。變數名稱不需要加引號。

先猜下方的結果：＿＿＿＿

````python
price = 30
quantity = 2
total = price * quantity
print(total)
````

結果為 `60`。Python 由上往下執行：先記住單價與數量，再計算總價，最後顯示結果。

**變數命名先守這三個習慣：**
- 用容易理解的英文字，例如 `price`；多字可用底線，例如 `cookie_price`。
- 不用空格，也不要用數字開頭。
- 大小寫要一致；`price` 和 `Price` 是不同名稱。不要把 `print` 拿來當變數名稱。

## 4｜Colab 實驗：改了文字，數值就更新了嗎？
下面是兩個分開的程式區。請照順序操作：

````python
# 設定單價
price = 30
````

````python
# 顯示金額
print(price * 2)
````

### 記下你的觀察
- 只修改文字、執行第二格時，我看到：＿＿＿＿
- 兩格依序重新執行後，我看到：＿＿＿＿

照題目操作，結果應分別為 `60` 和 `100`。編輯程式碼只是改了畫面上的文字，**執行才會更新運算環境中的值**。

如果一開始就出現 `NameError`，代表這個運算環境還沒設定 `price`；先執行第一格。

**還有一件事：** `total = price * quantity` 會保存「當下算出的值」，不會像試算表公式一樣自動跟著其他變數變動。改單價後，也要重新執行計算 `total` 的那一行。
